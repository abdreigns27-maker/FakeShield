## Step 1 — Imports

```python
import pickle
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score

print('Ready!')
```

## Step 2 — Load test data and all saved models

```python
X_test = pickle.load(open('X_test.pkl', 'rb'))
y_test = pickle.load(open('y_test.pkl', 'rb'))

model_files = {
    'Logistic Regression (Abdullah)': 'logistic_regression_model.pkl',
    'Naive Bayes (Tayyeba)'         : 'naive_bayes_model.pkl',
    'Random Forest (Danish)'        : 'random_forest_model.pkl',
    'SVM (Mohsin)'                  : 'svm_model.pkl',
}

models = {name: pickle.load(open(f, 'rb')) for name, f in model_files.items()}
print('Loaded', len(models), 'models.')
```

## Step 3 — Compare accuracy

```python
scores = {}
for name, model in models.items():
    acc = accuracy_score(y_test, model.predict(X_test))
    scores[name] = acc
    print(f'{name:35s} {acc*100:.2f}%')
```

## Step 4 — Accuracy bar chart

```python
plt.figure(figsize=(9, 5))
names = list(scores.keys())
accs  = [scores[n]*100 for n in names]
colors = ['#4C72B0', '#55A868', '#C44E52', '#8172B2']
bars = plt.bar(names, accs, color=colors, edgecolor='white')
plt.ylim(80, 100)
plt.ylabel('Accuracy (%)')
plt.title('FakeShield — Model Accuracy Comparison')
plt.xticks(rotation=15, ha='right')
for bar, acc in zip(bars, accs):
    plt.text(bar.get_x()+bar.get_width()/2, bar.get_height()+0.2,
             f'{acc:.2f}%', ha='center', fontweight='bold')
plt.tight_layout()
plt.savefig('accuracy_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
```

## Step 5 — Save the best model for deployment
The highest-accuracy model is saved as `best_model.pkl`. Upload this plus `tfidf_vectorizer.pkl` to the backend folder.

```python
best_name = max(scores, key=scores.get)
best_model = models[best_name]

pickle.dump(best_model, open('best_model.pkl', 'wb'))
print(f'Best model: {best_name} ({scores[best_name]*100:.2f}%)')
print('Saved as best_model.pkl  <- used by the Flask backend')
```
