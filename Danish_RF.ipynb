## Step 1 — Import libraries

```python
import pickle
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print('Libraries imported!')
```

## Step 2 — Load the shared preprocessed data
These files come from `preprocess.ipynb`. Every member uses the same data.

```python
X_train = pickle.load(open('X_train.pkl', 'rb'))
X_test  = pickle.load(open('X_test.pkl',  'rb'))
y_train = pickle.load(open('y_train.pkl', 'rb'))
y_test  = pickle.load(open('y_test.pkl',  'rb'))

print('Data loaded.')
print('Train shape:', X_train.shape)
print('Test shape :', X_test.shape)
```

## Step 3 — Train the Random Forest model
Random Forest builds many decision trees and combines their votes. May take a few minutes to train.

```python
model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
print('Training... please wait.')
model.fit(X_train, y_train)
print('Training complete!')
```

## Step 4 — Evaluate the model
We check accuracy and a full classification report (precision, recall, F1).

```python
y_pred = model.predict(X_test)
acc = accuracy_score(y_test, y_pred)

print(f'Accuracy: {acc * 100:.2f}%')
print()
print('Classification Report:')
print(classification_report(y_test, y_pred, target_names=['Real', 'Fake']))
```

## Step 5 — Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Oranges',
            xticklabels=['Real', 'Fake'],
            yticklabels=['Real', 'Fake'])
plt.title('Random Forest — Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.tight_layout()
plt.show()
```

## Step 6 — Save the trained model
This `.pkl` file gets uploaded to the GitHub repo.

```python
pickle.dump(model, open('random_forest_model.pkl', 'wb'))
print('Model saved as random_forest_model.pkl')
```
